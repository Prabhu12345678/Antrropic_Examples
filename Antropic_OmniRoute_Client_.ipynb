```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic

# Tunnelling commands worked fine on OS Ubuntu 26.04.01 running on Lenevo T520 Laptop
# Command installs local tunnelling dependencies : @machine> sudo npm install -g localtunnel
# Command exposes local Omniroute llm Collabrator site http://localhost:20128 : @machine> lt --port 20128
# This is a tunneled URL, as OmniRoute is running on local.
# Antropic Python Client code runs on Colab Server hence tunnelling is required
# to exporse local address of the OmniRoute Site

client = Anthropic(
    api_key=ANTHROPIC_AUTH_TOKEN,
    base_url='https://quiet-cougars-battle.loca.lt',
    timeout=60.0)
model = 'prabhu-stack'

#"ANTHROPIC_BASE_URL": "http://localhost:20128",
#"ANTHROPIC_AUTH_TOKEN": "sk-162a84b111e19fab-95545e-00eb1ff5",
#"ANTHROPIC_API_KEY": "",
#"ANTHROPIC_MODEL": "prabhu-stack",
```

```python
def add_user_message(messages, text):
  message = {"role":"user", "content":text}
  messages.append(message)

def add_assitant_message(messages, text):
  message = {"role":"assistant", "content":text}
  messages.append(message)

def chat(messages):
  message = client.messages.create(
      model=model,
      max_tokens=1000,
      messages=messages
      )
  return message.content[0].text
```

```python
system_prompt = """
You were a patient math tutoe,
guide students to solve math problems step by step,
rather solving the problems in one go
"""

def chat(messages):
  message = client.messages.create(
      model=model,
      max_tokens=1000,
      messages=messages,
      system=system_prompt
      )
  return message.content[0].text

messages = []

math_prob = "How do I solve 5x + 3 = 2 for x?"

add_user_message(messages, math_prob)

math_sol = chat(messages)

print("\n ", math_sol)
```

```python
def chat(messages, system=None):

  params = {
      "model" : model,
      "max_tokens" : 1000,
      "messages" : messages,
      "system" : system
  }

  if system:
    params['system'] = system


  message = client.messages.create(**params)
  return message.content[0].text

messages = []

system = """
You were a patient math tutor,
guide students to solve math problems step by step,
rather solving the problems in one go
"""

math_prob = "How do I solve 5x + 3 = 2 for x?"

add_user_message(messages, math_prob)

math_sol = chat(messages, system = system)

print("\n ", math_sol)
```

```python
def chat(messages, system=None):

  params = {
      "model" : model,
      "max_tokens" : 1000,
      "messages" : messages
  }

  if system:
    params['system'] = system


  message = client.messages.create(**params)
  return message.content[0].text

messages = []

# Note : Precise vs concise
system = "You are a Python Developer writes very concise code."
#system = "You are a Python Developer writes very precise code."

math_prob = "Write a code to remove duplicate letters in a string."

add_user_message(messages, math_prob)

math_sol = chat(messages, system = system)

print("\n ", math_sol)
```

```python
messages = []

add_user_message(messages, "You were a world food expert, tell me in two sentence, why Indian food is so fascinating?")

assistant_answer = chat(messages)

add_assitant_message(messages, assistant_answer)

print("\n " ,messages)

add_user_message(messages, "Write one more sentence")

assistant_answer = chat(messages)

add_assitant_message(messages, assistant_answer)

print("\n ",messages)
```
