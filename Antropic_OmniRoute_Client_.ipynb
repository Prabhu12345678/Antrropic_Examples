```python
from google.colab import userdata

# Get your API key from Colab secrets
#ANTHROPIC_API_KEY = userdata.get('claude_key')
ANTHROPIC_AUTH_TOKEN = userdata.get('auth_token')
```

```python
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic

# Tunnelling commands worked fine on OS Ubuntu 26.04.01 running on Lenevo T520 Laptop
# Command installs local tunnelling dependencies : @machine> sudo npm install -g localtunnel
# Command exposes local Omniroute llm Collabrator site http://localhost:20128 : @machine> lt --port 20128

client = Anthropic(
    api_key=ANTHROPIC_AUTH_TOKEN,
    # This is a tunneled URL, as OmniRoute is running on local.
    # Antropic Python Client code runs on Colab Server hence tunnelling is required
    # to exporse local address of the OmniRoute Site
    base_url='https://long-lies-peel.loca.lt',
    timeout=60.0)
model = 'prabhu-stack'

#"ANTHROPIC_BASE_URL": "http://localhost:20128",
#"ANTHROPIC_AUTH_TOKEN": "sk-162a84b111e19fab-95545e-00eb1ff5",
#"ANTHROPIC_API_KEY": "",
#"ANTHROPIC_MODEL": "prabhu-stack",
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "Why are Indian food is so famous?"
        }
    ]
)
```
